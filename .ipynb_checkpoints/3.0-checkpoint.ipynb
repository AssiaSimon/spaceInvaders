```python
### 3.0 - Score calculé

"""
@author: Alex, Assia
"""
import json

try:  # import as appropriate for 2.x vs. 3.x
    import tkinter as tk
    import tkinter.messagebox as tkMessageBox # module pour afficher des messages à l'utilisateur
    import time  # pour gérer le temps
except:
    import Tkinter as tk
    import tkMessageBox
    import time

class SpaceInvaders(object):
    """
    Main Game class
    """
    def __init__(self):
        # créer la fenêtre principale
        self.root = tk.Tk()
        self.root.title("Space Invaders")
        self.player = None

        # ajouter un label demandant nom
        label = tk.Label(self.root, text="Nom du joueur:")
        label.pack(pady=10)
        
        # ajouter un champ pour entrer le nom
        self.entree_nom = tk.Entry(self.root)  
        self.entree_nom.pack(pady=10)
        
        # ajouter un bouton pour récupérer le nom
        bouton = tk.Button(self.root, text="START GAME", command=self.recuperer_nom)
        bouton.pack(pady=10)
        
    def play(self):
        # Lancement de la boucle principale tkinter
        self.root.mainloop()

    def recuperer_nom(self):
        self.player = self.entree_nom.get()  # récupère le texte saisi dans le champ
        if self.player == "":
            self.player = "joueur" 
        print("Nom du joueur : " + self.player)

        # effacer les widgets de la fenêtre d'entrée
        for widget in self.root.winfo_children():
            widget.destroy()

        # lancer le jeu
        self.startGame()

    def startGame(self):
        # création frame pour labels (score, nom du joueur)
        self.frame_labels = tk.Frame(self.root, bg="black")
        self.frame_labels.pack(side="top", fill="both", padx=2, pady=2)

        # création du Frame pour jeu 
        self.frame = tk.Frame(self.root, bg="black")
        self.frame.pack(side="top", fill="both", expand=True)
        # widget doit "expanser" pour occuper tout l'espace disponible, en fonction des autres widgets dans son parent
        
        # création du game (commenté ici pour ne pas l'inclure tout de suite)
        self.game = Game(self.frame, self.frame_labels, self.player)
        self.game.start_animation()


class Game(object):
    def __init__(self, frame, frame_labels, player):
        self.frame = frame
        self.frame_labels = frame_labels
        self.gameOver = False   # initialiser gameOver à False
        self.fleet = Fleet(self)
        self.defender = Defender()
        self.height = 800
        self.width = 500 
        self.player = player
        self.score = 0

        # ajouter un label pour afficher le score
        self.score_label = tk.Label(self.frame_labels, text="Score : "+str(self.score), fg="#00FF00", bg="black", font=("Arial", 24))
        self.score_label.grid(row=0, column=0, padx=10, pady=5, sticky="w")

        # ajouter le nom du joueur 
        self.player_label = tk.Label(self.frame_labels, text=self.player, fg="#00FF00", bg="black", font=("Arial", 24))
        self.player_label.grid(row=0, column=1, padx=10, pady=5, sticky="e")  # Aligné à droite

        # configurer les colonnes pour qu'elles prennent autant d'espace que possible
        self.frame_labels.grid_columnconfigure(0, weight=1)  # Colonne 0 (score)
        self.frame_labels.grid_columnconfigure(1, weight=1)  # Colonne 1 (nom joueur)
        
        # création du canvas pour le jeu
        self.canvas = tk.Canvas(self.frame, width=self.width, height=self.height, bg='black')
        self.canvas.pack(fill="both", expand=True)

        # installation sur canvas de defender et fleet
        self.defender.install_in(self.canvas)
        self.fleet.install_in(self.canvas)

        # liaison clavier
        self.canvas.focus_set()   # définit le focus sur un widget, canvas recevra les événements du clavier
        self.canvas.bind("<KeyPress>", self.defender.keypress)

        # enregistrer le temps de départ
        self.start_time = time.time()

        
    def update_score(self):
        if not self.gameOver:
            temps_ecoule = time.time() - self.start_time
            self.score = int(temps_ecoule)
            self.score_label.config(text="Score : "+ str(self.score))
            self.canvas.after(1000, self.update_score)  # mettre à jour toutes les secondes
    
    def animation(self):
        if not self.gameOver:
            self.move_bullets()  # lance la boucle d'animation des bullets
            #self.update_score()  # commence à mettre à jour le score
            self.move_aliens_fleet()
            self.update_score()
   
    def start_animation(self):
        self.animation()

    def move_bullets(self):
        if self.gameOver:
            return # On arrête ici
        
        bullets_to_remove = []
    
        # itérer sur une copie de la liste
        for bullet in list(self.defender.fired_bullets):
                
            # coordonnées du bullet
            x1, y1, x2, y2 = self.canvas.coords(bullet.id) 
                
            # si le bullet sort en haut → on le détruit
            if y1 < 0:
                self.canvas.delete(bullet.id)
                bullets_to_remove.append(bullet)
            else :
                # deplacement vers le haut
                bullet.move_in(self.canvas)
    
        # suppression des bullets sortis de la liste du defender
        for bullet in bullets_to_remove:
            if bullet in self.defender.fired_bullets:
                self.defender.fired_bullets.remove(bullet)
    
        # vérification des collisions avec les aliens après chaque mouvement de bullet
        self.fleet.manage_touched_aliens_by(self.canvas, self.defender)
    
        # relancer l'animation périodiquement (10 ms)
        if not self.gameOver:
            self.canvas.after(10, self.move_bullets)
        
    def move_aliens_fleet(self):
        if not self.gameOver: 
            self.fleet.move_in(self.canvas)

            # relancer le mouvement toutes les 100 ms
            if not self.gameOver: 
                self.canvas.after(100, self.move_aliens_fleet)

class Bullet(object):
    def __init__(self, shooter):
        self.radius = 5
        self.color = "red"
        self.speed = 8
        self.id = None
        self.shooter = shooter  # objet Defender
        
    def install_in(self, canvas):
        # récupère position du shooter 
        x1, y1, x2, y2 = canvas.coords(self.shooter.id)

        # centre du shooter
        cx = (x1 + x2) / 2
        cy = y1

        # création du bullet
        self.id = canvas.create_oval(cx - self.radius, cy - self.radius, cx + self.radius, cy + self.radius, fill=self.color)
        
    def move_in(self, canvas): 
        canvas.move(self.id, 0, -self.speed)

class Defender(object):
    def __init__(self):
        self.width = 20
        self.height = 20
        self.move_delta = 20
        self.id = None
        self.max_fired_bullets = 8
        self.fired_bullets = []

    def install_in(self, canvas):
        # création du carré vert
        x, y = 250, 750
        self.id = canvas.create_rectangle(x, y, x + self.width, y + self.height, fill='gray', tags='defender')

    def move_in(self, canvas, dx):
        canvas.move(self.id, dx, 0)

    def fire(self, canvas):
        # On limite le nombre de bullets à max_fired_bullets
        if len(self.fired_bullets) < self.max_fired_bullets:

            # créer un bullet
            bullet = Bullet(self)

            # afficher le bullet dans le canvas
            bullet.install_in(canvas)

            # l’ajouter à la liste des bullets actifs
            self.fired_bullets.append(bullet)

    def keypress(self, event):
        canvas = event.widget
        # deplacement en tenant compte de canvas_width
        x1, y1, x2, y2 = canvas.coords(self.id) 
        canvas_width = canvas.winfo_width() # obtenir le wisdh réel du widget canvas

        # Déplacement vers la gauche
        if event.keysym == 'Left':
            nv_x1 = x1 - self.move_delta
            if nv_x1 < 0:
                canvas.coords(self.id, 0, y1, self.width, y2)
            else : 
                self.move_in(canvas, -self.move_delta)

        # Déplacement vers la droite
        elif event.keysym == 'Right':
            nv_x2 = x2 + self.move_delta
            if nv_x2 > canvas_width:
                canvas.coords(self.id, canvas_width - self.width, y1, canvas_width, y2)
            else: 
                self.move_in(canvas, self.move_delta)

        # Lance les bullets        
        elif event.keysym == 'space':
            # lance les tires
            self.fire(canvas)

class Fleet(object):
    def __init__(self, game):
        self.game = game  # Référence à l'instance de Game
        self.aliens_lines = 5
        self.aliens_columns = 10
        self.aliens_inner_gap = 20
        self.alien_x_delta = 20
        self.alien_y_delta = 15
        fleet_size = self.aliens_lines * self.aliens_columns
        self.aliens_fleet = [None] * fleet_size
        self.pim = tk.PhotoImage(file='alien.gif') 
        self.pim = self.pim.subsample(2, 2) # redimensionner l'image : diviser la taille par 2
        self.gameOver = self.game.gameOver
    
    def install_in(self, canvas):
        # créer la matrice d'aliens
        x0 = self.alien_x_delta
        y = self.alien_y_delta
        w = self.aliens_inner_gap

        idx = 0
        for ligne in range(self.aliens_lines) :
            x = x0   # réinitialiser x à chaque ligne
            for column in range(self.aliens_columns):
                alien = Alien()
                alien.install_in(canvas, x, y, img=self.pim , tag="alien")
                self.aliens_fleet[idx] = alien
                idx += 1
                x += w + self.aliens_inner_gap
            y += w + self.aliens_inner_gap   # descendre d'une ligne
        
    def move_in(self, canvas):
        if not self.gameOver: 
            canvas_width = canvas.winfo_width()
            canvas_height = canvas.winfo_height()
    
            touch_bord = False
    
            dy = 0
            dx = self.alien_x_delta 
            
            for alien in self.aliens_fleet:
                if alien.alive :
                    x1, y1, x2, y2 = canvas.bbox(alien.id)
                    if (x2 + self.alien_x_delta >= canvas_width) or (x1 + self.alien_x_delta <= 0):
                        touch_bord = True 
    
            # si on touche un bord on change la direction
            if touch_bord :
                self.alien_x_delta = -self.alien_x_delta
           
            # récupère les coordonnées de chaque alien avec bbox
            # déplacement 
            for alien in self.aliens_fleet:
                if alien.alive:
                    alien.move_in(canvas, dx, 0)
    
            # Vérification si arrivé à destination 
            x_def, y_def, x_def2, y_def2 = canvas.bbox("defender") 
    
            if touch_bord :
                for alien in self.aliens_fleet:
                    if alien.alive and y_def > y2: 
                        dy = self.aliens_inner_gap
                        alien.move_in(canvas, 0, dy)
                    else :
                        if not self.gameOver and y2 >= y_def: # s'affiche 1 seule fois 
                            tkMessageBox.showinfo("Game Over !")
                            self.gameOver = True  # Fin du jeu 
                            self.game.gameOver = True # mise à jour de gameOver dans Game
                
    def manage_touched_aliens_by(self, canvas, defender):
        bullets_to_remove = []

        # vérifier les collisions aliens / projectiles
        for alien in self.aliens_fleet:
            if alien.alive:     # Si l'alien est encore vivant
                for projectile in defender.fired_bullets:
                    if projectile.id is not None : 
                        if alien.touched_by(canvas, projectile): 
                            bullets_to_remove.append(projectile)  # Ajoute les bullets à supprimer

            else :     # Alien est mort
                self.aliens_fleet.remove(alien)  # supprimer l'alien de la flotte
        
        # supprimer les projectiles qui ont touché des aliens
        for bullet in bullets_to_remove:
            if bullet in defender.fired_bullets:
                defender.fired_bullets.remove(bullet)
                canvas.delete(bullet.id)  # supprimer le projectile du canvas

        # vérifier si les aliens sont morts
        if all(not alien.alive for alien in self.aliens_fleet): 
            if not self.gameOver : # si pas fin de jeu
                tkMessageBox.showinfo("You Win !")
                self.gameOver = True
                self.game.gameOver = True # mise à jour de gameOver dans Game
        

class Alien(object):
    def __init__(self):
        self.id = None
        self.alive = True
        self.explosion_image = None 
        
    def touched_by(self, canvas, projectile):
        if not self.alive : 
            return False # alien déjà mort
        
        x1p, y1p, x2p, y2p = canvas.bbox(projectile.id)
        x1, y1, x2, y2 = canvas.bbox(self.id)
        
        if (x2p >= x1 and x1p <= x2) and (y2p >= y1 and y1p <= y2):
            self.alive = False
            canvas.delete(self.id)  # Supprimer l'alien de l'écran
            
            # créer un effet d'explosion à la place de l'alien
            self.explosion_image = canvas.create_oval(x1, y1, x2, y2, fill="red", outline="yellow")
            # Attendre 300 ms avant de supprimer l'explosion
            canvas.after(100, canvas.delete, self.explosion_image)  
            
            return True # collision 
        return False # aucune collision
        
    def install_in(self, canvas, x, y, img=None, tag=None):
        self.id = canvas.create_image(x, y, image=img, tags=tag)
    
    def move_in(self, canvas, dx, dy):
        canvas.move(self.id, dx, dy)
        
# créer et ouvrir la fenêtre
fen = SpaceInvaders()
fen.play()
```
